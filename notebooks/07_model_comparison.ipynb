Most beginners stop after training a model.

Real ML engineers ask:

Which model is best and why?

Goal

Compare:

Linear Regression
Random Forest
TensorFlow Neural Network

using:

MAE
RMSE
R²

and decide the final production model.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest",
        "TensorFlow NN"
    ],
    
    "MAE": [
        2.4620,
        1.6744,
        1.9280
    ],
    
    "RMSE": [
        3.2561,
        2.3960,
        2.8030
    ],
    
    "R2": [
        0.7923,
        0.8875,
        0.8461
    ]
})

results
```

## Sort Models

```python
results.sort_values(
    by="R2",
    ascending=False
)
```

## MAE Comparison

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=results,
    x="Model",
    y="MAE"
)

plt.title("MAE Comparison")

plt.show()
```

## RMSE Comparison

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=results,
    x="Model",
    y="RMSE"
)

plt.title("RMSE Comparison")

plt.show()
```

## R² Comparison

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=results,
    x="Model",
    y="R2"
)

plt.title("R² Comparison")

plt.show()
```

## Combined Visualization

```python
fig, axes = plt.subplots(
    1,
    3,
    figsize=(15,5)
)

sns.barplot(
    data=results,
    x="Model",
    y="MAE",
    ax=axes[0]
)

sns.barplot(
    data=results,
    x="Model",
    y="RMSE",
    ax=axes[1]
)

sns.barplot(
    data=results,
    x="Model",
    y="R2",
    ax=axes[2]
)

axes[0].set_title("MAE")
axes[1].set_title("RMSE")
axes[2].set_title("R²")

plt.tight_layout()

plt.show()
```

## Best Model

```python
best_model = results.loc[
    results["R2"].idxmax()
]

best_model
```

# Final Conclusion

Models Compared:

Random Forest achieved:
