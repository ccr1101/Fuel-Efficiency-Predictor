```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/mpg.csv")
```

```python
df.info()
```

## Step 1: Inspect Missing Rows

```python
df[df["horsepower"].isnull()]
```

## Step 2: Handle Missing Values

```python
df = df.dropna()
df.isnull().sum()
```

## Step 4: Inspect Categorical Columns

```python
# check unique origins

df["origin"].unique()
```

## Step 5: Remove Car Name

```python
df = df.drop("name", axis=1)
df.head()
```

## Step 6: Encode Origin

```python
# Use One-Hot Encoding

df = pd.get_dummies(
    df,
    columns=["origin"],
    drop_first=True
)
```

## Step 7: Verify New Columns

```python
df.head()
```

## Step 8: Check Data Types

```python
df.info()

bool_cols = df.select_dtypes(include="bool").columns
df[bool_cols] = df[bool_cols].astype(int)
```

## Step 9: Separate Features and Target

```python
# Target
y = df["mpg"]

# Features
X = df.drop("mpg", axis=1)
```

## Step 10: Verify Shapes

```python
print("Features:", X.shape)
print("Target:", y.shape)
```

## Step 11: Save Clean Dataset

```python
# create a clean version

df.to_csv(
    "../data/mpg_clean.csv",
    index=False
)

# Verify
pd.read_csv("../data/mpg_clean.csv").head()
```
